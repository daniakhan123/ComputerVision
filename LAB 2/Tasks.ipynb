```python

import cv2
import numpy as np

img = cv2.imread(r'D:\Danias Code\CV\LAB 02\task1_chest_xray\PNEUMONIA\PNEUMONIA_2.png', 0)

equalized = cv2.equalizeHist(img)
cv2.imshow('Original Image', img)
cv2.imshow('Equalized Image', equalized)

heatmap = cv2.applyColorMap(equalized, cv2.COLORMAP_JET)
cv2.imshow('Heatmap', heatmap)
b, g, r = cv2.split(heatmap)
b = cv2.add(b, 10)
balanced = cv2.merge([b, g, r])
cv2.imshow('Balanced Image', balanced)

_, threshold = cv2.threshold(equalized, 150, 255, cv2.THRESH_BINARY)
cv2.imshow('Thresholded Image', threshold)
c = 255 / np.log(1 + np.max(img))
log_img = c * (np.log(img + 1))
log_img = np.uint8(log_img)

cv2.imshow('Log Transformed Image', log_img)
gamma = 0.5
gamma_img = 255 * (img / 255) ** gamma
gamma_img = np.uint8(gamma_img)

cv2.imshow('Gamma Corrected Image', gamma_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
